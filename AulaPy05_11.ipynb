```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

# Configuracao visual para gráficos
sns.set_theme(style="whitegrid")

# Criando o dataset sintetico (imoveis em Recife)
data = {
    "area_m2": [65, 120, 45, 80, 90, 150, 55, 110, 75, 130, 200, 105],
    "quartos": [2, 3, 1, 2, 2, 4, 1, 3, 2, 3, 4, 2],
    "vagas": [1, 2, 0, 1, 1, 2, 1, 2, 1, 3, 1, 2], # Ajustado para conter exatamente 12 elementos
    "dist_praia_m": [150, 50, 800, 300, 400, 100, 600, 200, 350, 80, 50, 250],
    "preco_real": [450000, 950000, 220000, 580000, 620000, 120000, 350000, 880000, 510000, 980000, 1800000, 750000]
}

df = pd.DataFrame(data)
print("Base de dados Carregada!")
df.head()
```

A regressao Linear Multipla busca encontra os melhores pesos para cada ingrediente da nossa formula:

(Intercepto): O custo base ou tráfego mínimo garantido.

(inclinacao/Peso): O quanto cada visualizacao ou metro quadradoaumenta o valorfinal.

(Residuo): O caos imprevisivel ou fatores fora da formula.

```python
# Definindo variáveies Independentes (x) e alvo (y)
x = df.drop("preco_real", axis=1)
y = df["preco_real"]

# Divisao de treino (80%) e teste (20%)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Instanciando e treinando
modelo = LinearRegression()
modelo.fit(x_train, y_train)

# Gerando Previsoes
previsoes = modelo.predict(x_test)
```

Avaliando o placar do Jogo para saber se nosso modelo é confiável, usamos três métricas principais:

MAE(Erro Medio Absoluto): Diz em média quantos reais erramos por cada imóvel.

RMSE(Raiz do erro quadratico médio):  Penaliza severamente desvios catastróficos.

R²(Coeficiente de determinação): Indica a porcentagem da realidade que conseguimos explicar.

```python
mae = metrics.mean_absolute_error(y_test, previsoes)
rmse = np.sqrt(metrics.mean_squared_error(y_test, previsoes))
r2 = metrics.r2_score(y_test, previsoes)

print(f"MAE (Reguá do GPS): R$ {mae: .2f}")
print(f"RMSE (Lupa de Outliers): R$ {rmse: .2f}")
print(f"R² : {r2: .2f}")
```
